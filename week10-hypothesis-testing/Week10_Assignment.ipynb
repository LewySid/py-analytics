```python
import numpy as np
import pandas as pd
from scipy import stats

sales = pd.read_csv("sales-data.csv")
products = pd.read_csv("product-performance.csv")
```

Task 1: Sales Performance Analysis Using Hypothesis Testing


One-Sample T Test on 2024 Sales:

Objective: Test if the average monthly sales in 2024 are significantly different from a benchmark of $650,000.

Data: Use the sales-data.csv file and filter for records from 2024.

Requirements:
- Calculate the mean monthly sales for 2024.
- Perform a one-sample t test comparing the sample mean to $650,000.
- Interpret the t statistic and p value.

```python
sales_2024 = sales[sales["Year"] == 2024]["Sales"]
benchmark = 650000

print("Mean 2024 Sales:", sales_2024.mean())

# Perform a one-sample t-test
t_stat, p_value = stats.ttest_1samp(sales_2024, benchmark)
print("T-Statistic:", t_stat)
print("P-value:", p_value)

# Confidence interval
ci = stats.t.interval(0.95, len(sales_2024) - 1, loc=sales_2024.mean(), scale=stats.sem(sales_2024))
print("95% CI:", ci)
```

Two-Sample T‑Test Comparing 2023 and 2024 Sales:

Objective: Determine if there is a statistically significant difference between average monthly sales in 2023 and 2024.

Data: Use the sales-data.csv file.

Requirements:
- Compute the average monthly sales for each year.
- Conduct a two-sample t‑test to compare the two means.
- Explain the business implications of your findings.

```python
sales_2023 = sales[sales["Year"] == 2023]["Sales"]

print("Mean 2023 Sales:", sales_2023.mean())
print("Mean 2024 Sales:", sales_2024.mean())

# Perform a two-sample t-test
t_stat, p_value = stats.ttest_ind(sales_2023, sales_2024)
print("T-Statistic:", t_stat)
print("P-value:", p_value)

# Confidence intervals
ci_2023 = stats.t.interval(0.95, len(sales_2023) - 1, loc=sales_2023.mean(), scale=stats.sem(sales_2023))
ci_2024 = stats.t.interval(0.95, len(sales_2024) - 1, loc=sales_2024.mean(), scale=stats.sem(sales_2024))
print("95% CI 2023:", ci_2023)
print("95% CI 2024:", ci_2024)
```

Task 2: Product Performance Evaluation


One-Sample T Test on Profit Margin:

Objective: Test if the average profit margin for all product categories is significantly different from a benchmark of 30%.

Data: Use the product-performance.csv file.

Requirements:
- Perform a one-sample t test against the 30% benchmark.
- Discuss what a significant result means for product strategy.

```python
margins = products["Profit_Margin"]
benchmark = 30

print("Mean Profit Margin:", margins.mean())

# Perform a one-sample t-test
t_stat, p_value = stats.ttest_1samp(margins, benchmark)
print("T-Statistic:", t_stat)
print("P-value:", p_value)

# Confidence interval
ci = stats.t.interval(0.95, len(margins) - 1, loc=margins.mean(), scale=stats.sem(margins))
print("95% CI:", ci)
```

Two-Sample Two-Sample T Test on ROI Between High and Low Revenue Categories:

Objective: Compare the average Return on Investment High Revenue and Low Revenue categories.

Note: You will need to define “High Revenue” and “Low Revenue” categories based on the data.

Data: Use the product-performance.csv file.

Requirements:
- Define a threshold to separate categories into “High Revenue” (e.g., Revenue > X) and “Low Revenue” (e.g., Revenue <= X). You can adjust this threshold if desired (try to get a relatively even split)
- Extract the ROI data for categories falling into each revenue group.
- Conduct an independent two-sample t-test to assess if there is a statistically significant difference in the average ROI between these two groups.
- Explain the potential business actions based on the result (e.g., Does ROI significantly differ based on the scale of the category? How might this influence investment or strategy?).

```python
threshold = products["Revenue"].median()
print("Threshold:", threshold)

high_rev = products[products["Revenue"] > threshold]
low_rev = products[products["Revenue"] <= threshold]
print("High Revenue:", list(high_rev["Category"]))
print("Low Revenue:", list(low_rev["Category"]))

roi_high = high_rev["ROI"]
roi_low = low_rev["ROI"]
print("Mean ROI High:", roi_high.mean())
print("Mean ROI Low:", roi_low.mean())

# Perform a two-sample t-test
t_stat, p_value = stats.ttest_ind(roi_high, roi_low)
print("T-Statistic:", t_stat)
print("P-value:", p_value)

# Confidence intervals
ci_high = stats.t.interval(0.95, len(roi_high) - 1, loc=roi_high.mean(), scale=stats.sem(roi_high))
ci_low = stats.t.interval(0.95, len(roi_low) - 1, loc=roi_low.mean(), scale=stats.sem(roi_low))
print("95% CI High:", ci_high)
print("95% CI Low:", ci_low)
```
