Assignment 4: Predicting Customer Lifetime Value with Multiple Regression


Setup

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

Task 1: Generate and Examine Data

```python
np.random.seed(110)
n_customers = 500

avg_purchase_value = np.random.normal(75, 25, n_customers)
purchase_frequency = np.random.normal(5, 2, n_customers)
customer_tenure = np.random.uniform(0.5, 10, n_customers)

avg_purchase_value = np.maximum(avg_purchase_value, 10)
purchase_frequency = np.maximum(purchase_frequency, 0.5)

clv = 150 + 10 * avg_purchase_value + 80 * purchase_frequency + 50 * customer_tenure + np.random.normal(0, 250, n_customers)
clv = np.maximum(clv, 50)

df_clv = pd.DataFrame({
    'AvgPurchaseValue': avg_purchase_value,
    'PurchaseFrequency': purchase_frequency,
    'CustomerTenure': customer_tenure,
    'CLV': clv
})

print(df_clv.head())
print(df_clv.describe())
```

Task 2: Define Features (X) and Target (y)

```python
X = df_clv[['AvgPurchaseValue', 'PurchaseFrequency', 'CustomerTenure']]
y = df_clv['CLV']
```

Task 3: Split Data into Training and Test Sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=110
)
print("Training set size:", X_train.shape[0])
print("Test set size:", X_test.shape[0])
```

Task 4: Create Model Instance

```python
model = LinearRegression()
```

Task 5: Train the Model


```python
model.fit(X_train, y_train)
```

Task 6: Print Intercept and Coefficients


```python
print(f"Intercept (β₀): {model.intercept_:.2f}")

coeffs_df = pd.DataFrame({'Feature': X.columns, 'Coefficient (β)': model.coef_})
print(coeffs_df)
```

Task 7: Evaluate on the Test Set

```python
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"R-squared (R²): {r2:.3f}")
print(f"RMSE: ${rmse:,.2f}")
```

Task 8: Predict CLV for a New Customer

```python
new_customer = pd.DataFrame({
    'AvgPurchaseValue': [85],
    'PurchaseFrequency': [6],
    'CustomerTenure': [4]
})

predicted_clv = model.predict(new_customer)
print(f"Predicted CLV: ${predicted_clv[0]:,.2f}")
```
