```python
# Logistic Regression for Banknote Authentication

# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import warnings

# Suppress potential convergence warnings for cleaner output (optional)
warnings.filterwarnings('ignore', category=FutureWarning)
warnings.filterwarnings('ignore', category=UserWarning)

# --- Reference Date ---
# Assignment context date: Saturday, April 5, 2025 (as per environment context)
print(f"Assignment starter code executed. Context Date: April 5, 2025")

# --- Task 1: Data Loading and Preparation ---
print("\n--- Task 1: Data Loading and Preparation ---")

# URL for the dataset
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt'

# Define column names
column_names = ['Variance', 'Skewness', 'Curtosis', 'Entropy', 'Class']

# Load data, specifying no header and comma separator
try:
    df = pd.read_csv(url, header=None, names=column_names, sep=',')
    print("Data loaded successfully.")
except Exception as e:
    print(f"Error loading data: {e}")
    # Exit or handle error appropriately in a real script
    exit()

# Display basic info - Verify data loaded correctly
print("\nDataFrame Head:")
print(df.head())
print("\nDataFrame Info:")
df.info()
print("\nDataFrame Description:")
print(df.describe())

# --- Prepare Features (X) and Target (Y) ---
# 1. Define the feature matrix X (should contain 'Variance', 'Skewness', 'Curtosis', 'Entropy')
X = df[['Variance', 'Skewness', 'Curtosis', 'Entropy']]
# 2. Define the target vector y (should contain the 'Class' column)
y = df['Class']
# 3. Describe the data (use pandas describe)
print("\nFeatures (X) Description:")
print(X.describe())
# Check the target distribution (genuine vs. forged)
print("\nClass Counts (0 = Genuine, 1 = Forged):")
print(y.value_counts())
print("\nClass Proportions:")
print(y.value_counts(normalize=True))

# --- Task 2: Train-Test Split ---
print("\n--- Task 2: Train-Test Split ---")
# 1. Split X and y into training and testing sets (X_train, X_test, y_train, y_test)
X_train, X_test, y_train, y_test = train_test_split(
# 2. Use test_size=0.20, random_state=42, and stratify=y
    X, y, test_size=0.20, random_state=42, stratify=y)
# 3. Print the shapes of the resulting arrays
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape:  {y_test.shape}")

# --- Task 3: Model Training ---
print("\n--- Task 3: Model Training ---")
# 1. Initialize a LogisticRegression model (with random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
# 2. Train the model using X_train_scaled and y_train
model = LogisticRegression(random_state=42)
model.fit(X_train_scaled, y_train)
print("Model training complete.")


# --- Task 4: Model Evaluation ---
print("\n--- Task 4: Model Evaluation ---")
# Your code here to:
# 1. Make predictions (y_pred) on the test data using the trained model
y_pred = model.predict(X_test_scaled)
# 2. Calculate the accuracy_score by comparing y_pred to y_test
accuracy = accuracy_score(y_test, y_pred)
# 3. Print the accuracy score (formatted nicely)
print(f"Accuracy: {accuracy:.2%}")
# 4. Add a print statement interpreting the accuracy in context
print(f"The model correctly classified {accuracy:.2%} of the banknotes in the test set as genuine or forged.")
# 5. Calculate the precision and recall scores.
from sklearn.metrics import precision_score, recall_score
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
print(f"\nPrecision: {precision:.2%}")
print(f"Recall: {recall:.2%}")
# 6. Add a print statement interpreting these scores in context
print(f"When the model flags a note as forged, it is correct {precision:.2%} of the time.")
print(f"The model caught {recall:.2%} of the actual forged notes in the test set.")
print("For a bank, missing a forged note costs more than a false alarm, so high recall matters most.")

# --- Task 5: Coefficient Interpretation ---
print("\n--- Task 5: Coefficient Interpretation ---")
# 1. Extract coefficients from the trained model
coefficients = pd.Series(model.coef_[0], index=X.columns)
# 2. Print the coefficients for 'Skewness' and 'Entropy'
print(f"Skewness coefficient: {coefficients['Skewness']:.4f}")
print(f"Entropy coefficient: {coefficients['Entropy']:.4f}")
print(f"Skewness odds ratio: {np.exp(coefficients['Skewness']):.4f}")
print(f"Entropy odds ratio: {np.exp(coefficients['Entropy']):.4f}")
print("Skewness: each 1 SD increase cuts the odds of a note being forged by about 99%.")
print("Entropy: each 1 SD increase raises the odds of a note being forged by about 29%.")

# --- Task 6: Conclusion ---
print("\n--- Task 6: Conclusion ---")
# Your code here to:
# 1. Write 1-2 print statements summarizing your findings (accuracy, interpretation highlights, model utility)
print(f"The model got {accuracy:.2%} of the banknotes right and caught every fake one.")
print("This could help the bank spot fake banknotes quickly, with a person double-checking any note it flags.")

# --- Task 7: Code Quality ---
print("\n--- Task 7: Code Quality ---")
# No code needed here, just ensure your script above is well-commented and clean.
print("Ensure code is commented, uses meaningful variable names, and runs without errors.")
```
